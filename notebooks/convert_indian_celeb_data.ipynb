```python
import re
import pandas as pd
from glob import glob
from pathlib import Path as p
```

```python
label_file = "../resources/Face-to-height-weight-BMI-estimation-/BMI data - Sheet1.csv"
```

```python
profile_df = pd.read_csv(label_file)
```

```python
profile_df.shape
```

```python
profile_df.head()
```

```python
data_folder = "../resources/Face-to-height-weight-BMI-estimation-/height_weight"
```

```python
all_files = glob(data_folder + "/*")

all_jpgs = sorted([img for img in all_files if ".jpg" in img or ".jpeg" in img or "JPG" in img])

print("Total {} photos ".format(len(all_jpgs)))
```

```python
def get_index_of_digit(string):
    match = re.search("\d", p(string).stem)
    return match.start(0)
```

```python
id_path = [(p(images).stem[:(get_index_of_digit(p(images).stem))], p(images).name, images) for images in all_jpgs]
```

```python
id_path[:5]
```

```python
image_df = pd.DataFrame(id_path, columns=['UID', 'file_name', 'path'])
```

```python
data_df = image_df.merge(profile_df)
```

```python
data_df.shape
```

```python
data_df.head()
```

```python
bmi_df = data_df[['file_name', 'BMI']].rename(columns={'file_name': 'name', "BMI": 'bmi'})
```

```python
bmi_df.shape
```

```python
bmi_df.head()
```

```python
bmi_df.to_csv('../data/data.csv', index=False)
```
